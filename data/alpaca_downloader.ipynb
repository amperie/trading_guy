################# Code to prepare barchart data for use ######################

```python
import pandas as pd

df = pd.read_csv("e:\\programming\\trading_guy\\data\\barchart\\GDXU.csv")
df_a = df.rename(columns={
    "Time":"timestamp", "Open": "open", "High": "high", "Low": "low", "Last": "close", "Volume": "volume",
    "Change": "trade_count"
    })
df_a['symbol'] = "GDXU"
df_a = df_a.drop("%Chg", axis=1)
df_a
```

```python
# Fix timestamp format to match required format: 2022-01-03 10:30:00+00:00
df_a['timestamp'] = pd.to_datetime(df_a['timestamp'], format='%Y-%m-%d %H:%M')
df_a['timestamp'] = df_a['timestamp'].dt.tz_localize('UTC')

print("Timestamp format fixed:")
print(df_a['timestamp'].head())
print(f"\nExample timestamp: {df_a['timestamp'].iloc[0]}")
```

```python
# Save the corrected dataframe with proper timestamp format
df_a.to_csv("e:\\programming\\trading_guy\\data\\barchart\\GDXU_fixed.csv", index=False)
print("Saved corrected data to GDXU_fixed.csv")
```
