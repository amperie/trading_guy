```python
from alpaca.data.historical import StockHistoricalDataClient
from alpaca.data.requests import StockBarsRequest
from alpaca.data.timeframe import TimeFrame
from datetime import datetime
import pandas as pd

# Initialize the client with authentication
client = StockHistoricalDataClient(
    api_key="PKQ56JJQ6LBTW25T76ZMXRIAN6",
    secret_key="CPCt567h6ZgPKxm49zFA3mgNGU2JK4HhaHdEvqdUUSxo"
)

# Define parameters
symbols = ["SPXU", "UPRO", "GDXU"]
start_date = datetime(2022, 1, 1)
end_date = datetime(2025, 11, 20)

# Create request
request_params = StockBarsRequest(
    symbol_or_symbols=symbols,
    timeframe=TimeFrame.Minute,  # Can be Minute, Hour, Day, Week, Month
    start=start_date,
    end=end_date,
    adjustment="split"
)

# Get the data
bars = client.get_stock_bars(request_params)

# Convert to DataFrame
df = bars.df

# If you want separate DataFrames per symbol:
for symbol in symbols:
    symbol_df = df.loc[symbol] if symbol in df.index.get_level_values(0) else pd.DataFrame()
    print(f"\n{symbol}:")
    print(symbol_df.head())
```
