```python
import mlflow

TRACKING_URI = "http://hp.lan:8899"
EXPERIMENT_NAME = "walk_forward_runs"

mlflow.set_tracking_uri(TRACKING_URI)

experiment = mlflow.get_experiment_by_name(EXPERIMENT_NAME)
if experiment is None:
    raise RuntimeError(f"Experiment not found: {EXPERIMENT_NAME}")

print(f"Experiment ID: {experiment.experiment_id}")

with mlflow.start_run(experiment_id=experiment.experiment_id, run_name="mlflow_sanity_test"):
    mlflow.log_param("sanity_check", True)
    mlflow.log_metric("test_metric", 1.0)
    print("Run created successfully")
```
